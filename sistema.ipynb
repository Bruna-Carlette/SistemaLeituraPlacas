```python
import os
import shutil

base = "/content/drive/MyDrive/grupo 6/TREINONOVA"

images_dir = os.path.join(base, "images")
labels_dir = os.path.join(base, "labels")

# cria pastas caso não existam
os.makedirs(images_dir, exist_ok=True)
os.makedirs(labels_dir, exist_ok=True)

for file in os.listdir(base):
    full_path = os.path.join(base, file)

    # pula se for diretório
    if os.path.isdir(full_path):
        continue

    # extensões de imagem
    if file.lower().endswith((".jpg", ".jpeg", ".png")):
        shutil.move(full_path, os.path.join(images_dir, file))

    # extensões de label YOLO
    elif file.lower().endswith(".txt"):
        shutil.move(full_path, os.path.join(labels_dir, file))

print("✔ Imagens e labels separados com sucesso!")
```
